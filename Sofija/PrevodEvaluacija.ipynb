```python
!pip install sacrebleu -q

from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from sacrebleu.metrics import BLEU, CHRF
import torch

dataset = load_dataset("Helsinki-NLP/opus-100", "en-mk")
n = 100
english_sentences = [item["translation"]["en"] for item in dataset["test"]][:n]
reference_macedonian = [item["translation"]["mk"] for item in dataset["test"]][:n]

device = "cuda" if torch.cuda.is_available() else "cpu"
model_name = "facebook/nllb-200-distilled-600M"
tokenizer = AutoTokenizer.from_pretrained(model_name, src_lang="eng_Latn")
model = AutoModelForSeq2SeqLM.from_pretrained(model_name).to(device)

def translate_to_macedonian(text):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)
    translated = model.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids("mkd_Cyrl"),
        max_length=512
    )
    return tokenizer.batch_decode(translated, skip_special_tokens=True)[0]

print("Преведувам...")
my_translations = []
for i, sentence in enumerate(english_sentences):
    my_translations.append(translate_to_macedonian(sentence))
    if i % 10 == 0:
        print(f"  {i}/{n} готово")
print("Завршено!")

bleu = BLEU()
chrf = CHRF()
bleu_score = bleu.corpus_score(my_translations, [reference_macedonian])
chrf_score = chrf.corpus_score(my_translations, [reference_macedonian])

print(f"\n📊 Резултати:")
print(f"BLEU Score:  {bleu_score.score:.2f}")
print(f"chrF Score:  {chrf_score.score:.2f}")

for i in range(5):
    print(f"\n--- Пример {i+1} ---")
    print(f"🇬🇧 EN: {english_sentences[i]}")
    print(f"✅ MK: {reference_macedonian[i]}")
    print(f"🤖 Превод: {my_translations[i]}")
```
